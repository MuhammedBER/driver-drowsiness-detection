```python
import os
import shutil
import glob
from tqdm import tqdm

Raw_DIR= r'C:\Users\Muhammad\Desktop\mrlEyes_2018_01'
for dirpath, dirname, filenames in os.walk(Raw_DIR):
    for i in tqdm([f for f in filenames if f.endswith('.png')]):
        if i.split('_')[4]=='0':
            shutil.copy(src=dirpath+'/'+i, dst=r'C:\Users\Muhammad\Desktop\the_data\Close Eyes')
        
        elif i.split('_')[4]=='1':
            shutil.copy(src=dirpath+'/'+i, dst=r'C:\Users\Muhammad\Desktop\the_data\Open Eyes') 

import random
def create_test_closed(source, destination, percent):
    
    path, dirs, files_closed = next(os.walk(source))
    file_count_closed = len(files_closed)
    percentage = file_count_closed * percent
    to_move = random.sample(glob.glob(source + "/*.png"), int(percentage))

    for f in enumerate(to_move):
        if not os.path.exists(destination):
            os.makedirs(destination)
        shutil.move(f[1], destination)
    print(f'moved {int(percentage)} images to the destination successfully.')  

def create_test_open(source, destination, percent):
    
    path, dirs, files_open = next(os.walk(source))
    file_count_open = len(files_open)
    percentage = file_count_open * percent
    to_move = random.sample(glob.glob(source + "/*.png"), int(percentage))

    for f in enumerate(to_move):
        if not os.path.exists(destination):
            os.makedirs(destination)
        shutil.move(f[1], destination)
    print(f'moved {int(percentage)} images to the destination successfully.')

create_test_closed(r'C:\Users\Muhammad\Desktop\the_data\Train\closed', 
                    r'C:\Users\Muhammad\Desktop\the_data\Test\closed', 
                    0.2)

create_test_open(r'C:\Users\Muhammad\Desktop\the_data\Train\open', 
                    r'C:\Users\Muhammad\Desktop\the_data\Test\open', 
                    0.2)
```

```python
import os

def count_images_in_directory(directory):
    image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.gif']
    image_count = 0

    for root, dirs, files in os.walk(directory):
        for file in files:
            if any(file.lower().endswith(ext) for ext in image_extensions):
                image_count += 1

    return image_count

dataset_directory = r'C:\Users\Muhammad\Desktop\the_data'
total_images = count_images_in_directory(dataset_directory)
print(f'Total number of images in the dataset: {total_images}')
```
